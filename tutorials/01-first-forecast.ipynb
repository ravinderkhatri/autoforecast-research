## Objective

Run a small forecasting comparison and find its saved results.

Install the package with the README commands and select that Python environment as the notebook kernel. Python 3.12/3.13 and Java 17 are required.

This example has **two products, 42 days and 84 observations**. No products are excluded. Each seven-day forecast has 14 values. We compare two simple included models. This teaches the process, not which model is best for your data.

All generated files stay in a separate workspace. Set `AUTOFORECAST_TUTORIAL_DIR` before starting the kernel to choose another folder.


## Step 1: Choose a place for the example files

We will keep data and results outside the library folder. The printed path shows where to find them. Set the environment variable only if you want a different folder.


```python
# Import the tools used in this step.
import os
from pathlib import Path

# Choose the workspace without changing the library checkout.
work = (
    Path(os.environ.get("AUTOFORECAST_TUTORIAL_DIR", Path.home() / "autoforecast-tutorials"))
    .expanduser()
    .resolve()
)
# Create the folder if this is your first visit.
work.mkdir(parents=True, exist_ok=True)
print(work)
```

### Understand the output

You should see the full workspace path. Open that folder to find the files created in these lessons. This path is a folder, not a forecast result.


Create the example once. The folder is kept when you rerun this cell.


## Step 2: Create and check the example data

We will create the two-product example, then freeze its data and evaluation rules. The helper runs the same commands shown in the README and stops if a command fails.


```python
# Import the tools used in this step.
import subprocess
import sys

# Keep the command-line example in its own folder.
project = work / "first-forecast"


# Stop with an error if an example command fails.
def cli(*args):
    subprocess.run([sys.executable, "-m", "autoforecast.cli", *map(str, args)], check=True)


# Create the sample files only on the first run.
if not project.exists():
    cli("init-example", project)
# Check and freeze the example data and evaluation dates.
cli("freeze", "--project-root", project, "--source", "data")
```

### Understand the output

The command should print a path ending in `benchmark.json`. This file identifies the checked study. Keep it with the example files. An exception means the command did not finish; read that error before continuing.


A **study** is a saved copy of the data and rules used to compare models. Find its manifest, then run the comparison. Repeating this campaign resumes matching saved work.


## Step 3: Run the comparison

We will find the one saved study and compare the included models. Read the result path printed by the command.


```python
# Find the saved study before starting its model comparison.
manifests = list(project.glob("**/benchmark.json"))
assert len(manifests) == 1, "Use the manifest printed by freeze if you have made several studies."
cli("run", "--project-root", project, "--benchmark", manifests[0])
```

### Understand the output

The command should print the campaign results folder. It contains saved model results. Each seven-day forecast has 14 values: two products times seven dates. A repeated call resumes matching saved work rather than starting the campaign again.
